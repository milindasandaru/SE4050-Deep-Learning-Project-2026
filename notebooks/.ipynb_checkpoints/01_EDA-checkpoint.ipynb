## 1. Load the Dataset

The PhiUSIIL Phishing URL Dataset is obtained from the UCI Machine Learning
Repository. It contains website and URL-related features that can be used to
classify websites as legitimate or phishing.

The raw dataset is loaded without modifying the original data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

print("Libraries imported successfully.")
```

```python
# Get the raw data directory
raw_dir = Path("../data/raw")

# Get the CSV file
csv_file = list(raw_dir.glob("*.csv"))[0]

# Load the dataset
df = pd.read_csv(csv_file)

print("Dataset loaded successfully.")
print(df.head())
```

## 4. Descriptive Statistics of Numerical Features

Descriptive statistics are used to examine the distribution, scale and
variability of numerical features before preprocessing.

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

print("Number of numerical columns:", len(numeric_columns))
```

```python
df[numeric_columns].describe().T
```

```python
feature_columns = df.columns.drop("label")

numeric_features = df[feature_columns].select_dtypes(
    include=["int64", "float64"]
).columns

zero_variance_features = [
    column for column in numeric_features
    if df[column].nunique() <= 1
]

print("Zero-variance numerical features:")
print(zero_variance_features)
```

```python
range_summary = pd.DataFrame({
    "Minimum": df[numeric_features].min(),
    "Maximum": df[numeric_features].max(),
    "Mean": df[numeric_features].mean(),
    "Std": df[numeric_features].std()
})

range_summary
```

```python
correlations = (
    df[numeric_features.tolist() + ["label"]]
    .corr()["label"]
    .drop("label")
    .sort_values(key=abs, ascending=False)
)

correlations
```

## 6. Visual Exploratory Data Analysis

### 6.1 Class Distribution

The target variable is `label`, where 0 represents one class and 1 represents the other class. The class distribution is visualized to determine whether the dataset is balanced or imbalanced.

```python
plt.figure(figsize=(5, 3))

ax = sns.countplot(data=df, x="label", hue="label", legend=False)

plt.title("Class Distribution")
plt.xlabel("Label")
plt.ylabel("Number of Samples")

# Add counts above bars
for container in ax.containers:
    ax.bar_label(container, fmt="%d")

plt.show()
```

```python
class_percentages = df["label"].value_counts(normalize=True).sort_index() * 100

plt.figure(figsize=(5, 3))

ax = sns.barplot(
    x=class_percentages.index,
    y=class_percentages.values,
    hue=class_percentages.index,
    legend=False
)

plt.title("Class Distribution (%)")
plt.xlabel("Label")
plt.ylabel("Percentage")

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f%%")

plt.show()
```

### 6.2 Numerical feature distributions

```python
url_features = [
    "URLLength",
    "DomainLength",
    "URLSimilarityIndex",
    "CharContinuationRate",
    "TLDLegitimateProb",
    "URLCharProb",
    "NoOfSubDomain",
    "NoOfLettersInURL",
    "NoOfDegitsInURL",
    "LetterRatioInURL",
    "DegitRatioInURL"
]

df[url_features].hist(
    figsize=(16, 12),
    bins=30,
    edgecolor="black"
)

plt.suptitle("Distributions of URL-Related Features", fontsize=16)
plt.tight_layout()
plt.show()
```

### 6.3 Website/content features

```python
web_features = [
    "LineOfCode",
    "LargestLineLength",
    "DomainTitleMatchScore",
    "URLTitleMatchScore",
    "NoOfImage",
    "NoOfCSS",
    "NoOfJS",
    "NoOfSelfRef",
    "NoOfEmptyRef",
    "NoOfExternalRef"
]

df[web_features].hist(
    figsize=(16, 12),
    bins=30,
    edgecolor="black"
)

plt.suptitle("Distributions of Website-Related Features", fontsize=16)
plt.tight_layout()
plt.show()
```

### 6.4 Binary feature distributions

```python
binary_features = [
    "IsDomainIP",
    "HasObfuscation",
    "IsHTTPS",
    "HasTitle",
    "HasFavicon",
    "Robots",
    "IsResponsive",
    "HasDescription",
    "HasExternalFormSubmit",
    "HasSocialNet",
    "HasSubmitButton",
    "HasHiddenFields",
    "HasPasswordField",
    "Bank",
    "Pay",
    "Crypto",
    "HasCopyrightInfo"
]

binary_percentages = df[binary_features].mean().sort_values(ascending=False) * 100

plt.figure(figsize=(7, 5))

sns.barplot(
    x=binary_percentages.values,
    y=binary_percentages.index,
    hue=binary_percentages.index,
    legend=False
)

plt.title("Percentage of Samples with Binary Features Present")
plt.xlabel("Percentage (%)")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

### 6.5 Correlation heatmap

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.2
)

plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

### 6.6 Correlations with the target

```python
target_corr = (
    numeric_df.corr()["label"]
    .drop("label")
    .sort_values()
)

top_corr = pd.concat([
    target_corr.head(10),
    target_corr.tail(10)
])

plt.figure(figsize=(8, 6))

sns.barplot(
    x=top_corr.values,
    y=top_corr.index,
    hue=top_corr.index,
    palette="coolwarm",
    legend=False
)

plt.axvline(0, color="black", linewidth=1)

plt.title("Features with Strongest Correlation with Label")
plt.xlabel("Correlation with Label")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

### 6.7 Compare important features

```python
important_features = [
    "URLSimilarityIndex",
    "IsHTTPS",
    "DomainTitleMatchScore",
    "URLTitleMatchScore",
    "HasSocialNet",
    "HasCopyrightInfo"
]

for feature in important_features:
    plt.figure(figsize=(5, 3))
    
    sns.boxplot(
        data=df,
        x="label",
        y=feature,
        hue="label",
        legend=False
    )
    
    plt.title(f"{feature} by Class")
    plt.xlabel("Label")
    plt.ylabel(feature)
    
    plt.tight_layout()
    plt.show()
```
